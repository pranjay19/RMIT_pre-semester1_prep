```python
import pandas as pd
```

```python
dic={
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona", "George", "Hannah", "Ian", "Julia"],
    "Age": [19, 20, 19, 21, 20, 22, 19, 20, 21, 19],
    "Major": ["Computer Science", "Mathematics", "Physics", "Biology", "Computer Science", 
              "Chemistry", "Mathematics", "Biology", "Physics", "Computer Science"],
    "GPA": [3.8, 3.2, 3.9, 3.5, 3.1, 4.0, 2.9, 3.7, 3.4, 3.9]
}
```

```python
df=pd.DataFrame(dic)
```
