# NumPy Arrays

| Feature | Python List | NumPy Array |
| :--- | :--- | :--- |
| Data type | Mixed allowed | Same data type only |
| Performance | Slower | Much faster (C-level) |
| Memory efficiency | Low | High |
| Vector operations | Manual loops | Easy + fast |

```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5])
```

```python
arr
```

<h1>Random generation function</h1>

```python
np.random.rand(10)  #numbers will lie in 0 - 1 (Normalisation concept)
```

```python
np.random.randn(10)  #numbers will lie in -3 - 3 (standardisation concept)
```

```python
np.random.randint(10)
```

```python
np.random.randint(10,20,10)
```
